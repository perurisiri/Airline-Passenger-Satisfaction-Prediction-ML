## Outlier Detection

```python
import pandas as pd
numeric_cols = df.select_dtypes(include='number').columns
result = []

for col in numeric_cols:
    Q1    = df[col].quantile(0.25)
    Q3    = df[col].quantile(0.75)
    IQR   = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    count = df[(df[col] < lower) | (df[col] > upper)].shape[0]
    result.append([col, count])

outlier_table = pd.DataFrame(result, columns=['Column', 'Outlier Count'])
print(outlier_table)
```

## Outlier Handling using Capping (IQR Method)

```python
for col in numeric_cols:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1

    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    
    df[col] = df[col].apply(lambda x: lower if x < lower else upper if x > upper else x)
```

## Segregate the data into features and target

```python
X = df.drop('satisfaction', axis = 1)
y = df['satisfaction']
X.shape, y.shape
```

## Split the data into training and testing

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size = 0.8, random_state = 42)
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

## Data Preprocessing

```python
from sklearn.preprocessing import OneHotEncoder
ohe=OneHotEncoder(sparse_output=False ,drop = 'first').set_output(transform = 'pandas')
ohe
```

```python
from sklearn.preprocessing import MinMaxScaler
minmax = MinMaxScaler().set_output(transform = 'pandas')
minmax
```

```python
from sklearn.preprocessing import RobustScaler
scaler = RobustScaler().set_output(transform = 'pandas')
scaler
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df['satisfaction'] = le.fit_transform(df['satisfaction'])
```

```python
from sklearn.compose import ColumnTransformer
ct = ColumnTransformer(
    transformers=[
        ('OHE', ohe, ['gender', 'customer_type', 'type_of_travel', 'class']),
        ('Robust', scaler, ['flight_distance', 'departure_delay', 'arrival_delay']),
        ('MINMAX',minmax,['age'])
    ],
    remainder='passthrough'
).set_output(transform='pandas')
```

```python
x_train_transformed = ct.fit_transform(X_train)
x_test_transformed = ct.transform(X_test)
```

## Baseline Model – Logistic Regression

```python
y_train = y_train.map({
    'Neutral or Dissatisfied': 0,
    'Satisfied': 1
})

y_test = y_test.map({
    'Neutral or Dissatisfied': 0,
    'Satisfied': 1
})
```

```python
import warnings
warnings.filterwarnings('ignore')
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

lr = LogisticRegression(max_iter=1000)
lr.fit(x_train_transformed, y_train)
```

```python
y_pred = lr.predict(x_test_transformed)
y_pred
```

```python
print("Accuracy:", accuracy_score(y_test, y_pred))
```

### Baseline Model Observation:

Logistic Regression achieved an accuracy of 86% on the test dataset. 
This indicates that the model correctly classified 86% of passenger satisfaction outcomes. 
The result serves as a baseline for comparing more advanced classification models such as Decision Tree, 
Random Forest, and XGBoost.

 ## Build ML Pipeline 

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, RobustScaler
from sklearn.ensemble import RandomForestClassifier

# Columns
cat_cols = ['type_of_travel','class']
num_cols = ['age','flight_distance'] 

# Preprocessing
preprocessor = ColumnTransformer([
    ('ohe', OneHotEncoder(handle_unknown='ignore'), cat_cols),
    ('scale', RobustScaler(), num_cols)
])

# Pipeline
pipeline = Pipeline([
    ('preprocessing', preprocessor),
    ('model', RandomForestClassifier(n_estimators=100, random_state=42))
])

# Train
pipeline.fit(X_train, y_train)

# Predict
y_pred = pipeline.predict(X_test)  
```
